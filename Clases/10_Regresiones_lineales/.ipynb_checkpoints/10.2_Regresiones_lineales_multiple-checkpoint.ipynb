**Regresión lineal múltiple**

<font color = 'Black'>
<font size = 3>

En la **regresión lineal simple**, estudiamos si una **variable dependiente** $Y$ depende linealmente de una **variable independiente** o de control $X$.

En la práctica, dicha situación rara vez se da, ya que la **variable dependiente** suele depender de mas un control (depende de más de una columna). Por lo tanto, en esta sección vamos a generalizar todo el estudio que hemos hecho para la **regresión lineal simple** al caso en que tengamos $k$ variables de control $X_1,X_2,\cdots,X_k$ ($k$ columnas **predictoras**). Es decir, una **variable dependiente** $Y$ y $k$ variables de control (si se tienen al menos dos **variables dependientes**, en la práctica lo que se hace es establecer al menos dos modelos: uno por **variable dependiente**).

Suponemos ahora que nuestro modelo es de la forma:

$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \cdots \beta_k X_k + \varepsilon$$ 

Este nuevo modelo ya no es posible graficarlo. Solamente, si tuviéramos 2 **columnas explicativas** y una de **control**, todavía se podría graficar en $\mathbb{R}^{3}$, donde se observarían planos y no rectas como en la **regresión lineal simple**. Cuando tenemos muchas más columnas ya no podrá ser visual el tratado pero, ese modelo es la ecuación de un hiperplano que vive en $\mathbb{R}^{k+1}$ dimensiones, $k$ **explicativas** y una **objetivo**. Es decir, $k + 1$ variables.

Los valores $\beta_0,\beta_1,...,\beta_k$ son los llamados parámetros de la **regresión** y se tienen que estimar a partir de una **muestra** las variables consideradas: $\{(x_{i1},x_{i2},...,x_{ik},y_i)\}_{i=1}^n$. En este caso, todas las $x$ son columnas, la $y$ es la variable objetivo y la $i$ es el número de fila.

Recordar que $k$ es el número de columnas y la $n$ es el número de filas.

Para que estas estimaciones se pueden realizar hay que suponer que $n > k$ (sup. tienes más filas que columnas), ya que en caso contrario tendríamos un problema subestimado: tendríamos más parámetros que valores en la **muestra**. Esto es para que el sistema de ecuaciones del problema tenga solución.

Denotamos por $\boldsymbol{x_i}$ a los valores de las $k$ variables independientes (columnas de control o explicativas) correspondientes al $i$-ésimo elemento de la **muestra**. Esto es: $\boldsymbol{x_i}=(x_{i1},x_{i2},\cdots,x_{ik})$. La columna $Y$ es la columna objetivo. Para cada columna $X$ se debe calcular el coeficiente $\beta$ correspondiente y además, el $\beta_{0}$ que es el intercepto. Por lo tanto, si tienes $k$ columnas explicativas vas a tener que calcular $k + 1$ coeficientes.

A partir de la **muestra** $\{(\boldsymbol{x_i},y_i)\}_{i=1}^n$ vamos a obtener estimaciones $b_0,b_1,\cdots,b_k$ para los parámetros $\beta_1,\beta_2,\cdots,\beta_k$.

Una vez obtenidas estas estimaciones, podemos definir los valores siguientes:

$$\begin{array}{l}\widehat{y_i}=b_0+b_1x_{i1}+...+b_kx_{ik},\\y_i=b_0+b_1x_{i1}+...+b_kx_{ik}+e_i\end{array}$$

Para simplificar la notación, escribimos los datos de la **muestra** en forma matricial. En primer lugar, definimos los **vectores** siguientes:

$$\boldsymbol{y}=\left(\begin{array}{c}y_1\\y_2\\\vdots\\y_n\end{array}\right),\,\boldsymbol{b}=\left(\begin{array}{c}b_0\\b_1\\\vdots\\b_k\end{array}\right),\,\boldsymbol{\widehat{y}}=\left(\begin{array}{c}\widehat{y_1}\\\widehat{y_2}\\\vdots\\\widehat{y_n}\end{array}\right),\,\boldsymbol{e}=\left(\begin{array}{c}e_1\\e_2\\\vdots\\e_n\end{array}\right),\,$$


Definimos $\boldsymbol{\mathrm{X}}$ a partir de los **datos** de la **muestra** de las **variables** $\boldsymbol{X}_i$:
$$\boldsymbol{\mathrm{X}}=\left(\begin{array}{ccccc}1&x_{11}&x_{12}&\cdots&x_{1k}\\1&x_{21}&x_{22}&\cdots&x_{2k}\\\vdots&\vdots&\vdots&\vdots&\vdots\\1&x_{n1}&x_{n2}&\cdots&x_{nk}\end{array}\right)$$

Por lo tanto $\boldsymbol{\widehat{y}}=\boldsymbol{\mathrm{X}}\boldsymbol{b}$ y $\boldsymbol{y}=\boldsymbol{\mathrm{X}\boldsymbol{b}}+\boldsymbol{e}$.

---

<font color = 'DodgerBlue'>
<font size = 5>
    
**Estimación de los parámetros por mínimos cuadrados.**

<font color = 'Black'>
<font size = 3>

Definimos el error cuadrático $SS_E$ como $$SS_E = \sum_{i=1}^ne_i = \sum_{i=1}^n(y_i-\widehat{y_i})^2 = \sum_{i=1}^n(y_i-(b_0+b_1x_{i1}+...+b_kx_{ik}))^2 = \|\boldsymbol{y}-\boldsymbol{\mathrm{X}}\boldsymbol{b}\|^2$$

El error cometido es $\|\boldsymbol{y}-\boldsymbol{\mathrm{X}}\boldsymbol{b}\|^2$, el cual se busca minimizar. La ventaja que se tiene es que $\mathrm{X}$ es una transformación lineal, $\boldsymbol{b}$ es un vector. Por lo tanto, $\mathrm{X} \boldsymbol{b}$ está en la imagen de esa transformación. Recordar que la imagen de una transformación lineal siempre es un espacio vectorial

Los estimadores $b_0,b_1,\cdots,b_k$ que buscamos son los que minimicen a $SS_E$.

<font color = 'Black'>
<font size = 3>

**Teorema.** 
Los estimadores por el método de los mínimos cuadrados de los parámetros $\{\beta_i\}$ a partir de la muestra $(\boldsymbol{x_i},y_i)_{i=1}^n$, son los siguientes: $$\boldsymbol{b}=(\boldsymbol{\mathrm{X}}^T\boldsymbol{\mathrm{X}})^{-1}( \boldsymbol{\mathrm{X}}^T\boldsymbol{y})$$ (ver el pdf [DemostracionMS](https://github.com/scidatmath2020/Inferencia-Estadistica/blob/master/DemostracionMS.pdf) en el repositorio de SCIDATA).

- **Observaciones del Teorema:**
    - $\boldsymbol{b}$ son tus estimaciones ($b_{0}$, $b_{1}$, $...$, $b_{k}$).
    - Del lado derecho de la ecuación conoces a todos. $X$ es tu columna de **datos** (variables independientes).
    - $(\boldsymbol{\mathrm{X}}^T \boldsymbol{\mathrm{X}})^{-1}$ existe y es invertible por la condición $n > k$. Como hay más filas que columnas, $(\boldsymbol{\mathrm{X}}^T \boldsymbol{\mathrm{X}})^{-1}$ tiene inversa.
    - $y$ son los valores de la columna objetivo.

Para la prueba del teorema anterior se necesitan otros teoremas y resultados importantes de **análisis funcional**, los cuales se listan a continuación.

**Teorema 1.** Sean $H$ un espacio con producto interno y $\emptyset \neq M \subseteq H$ un subconjunto convexo y completo. Entonces, para cada $u \in H$ existe un único $\tilde{u} \in M$ tal que

$$\lVert u - \tilde{u} \rVert = \inf_{v \in M} \lVert u - v \rVert$$ 

**Teorema 2.** Bajo la misma noción que en el teorema anterior, si $M$ es un subespacio cerrado de $H$, entonces para cada $u \in H$ se cumple $u - \tilde{u} \in M^{\perp}$.

**Teorema 3.** si $T \in \mathbb{B}(H, K)$, donde $H$ y $K$ son de Hilbert, entonces 

$$Im(T)^{\perp} = Ker(T^{\ast}) \text{ y } \overline{Im(T^{\ast})} = \overline{Im(T^{\ast}T)}$$

**Corolario.** Los estimadores $\boldsymbol{b} = (b_{0}, b_{1}, ... , b_{k})$ por el método de los mínimos cuadrados de los parámetro $\{ \beta_{i} \}^{k}_{i = 0}$ a partir de la muestra $\{ (\boldsymbol{x_{i}}, y_{i}) \}^{n}_{i = 1}$ se calculan como

$$\boldsymbol{b} = (\mathrm{X}^T \mathrm{X})^{-1} \mathrm{X}^{T} y.$$

*Demostración.* 

Observemos que queremos hallar $\boldsymbol{b} \in \mathbb{R}^{k + 1}$ que minimice a $\| \boldsymbol{y} - \boldsymbol{\mathrm{X}} \boldsymbol{b}\|^2$. Por los Teoremas 1 y 2, tomando $M = Im(X)$, tenemos $\tilde{\boldsymbol{y}} = X \boldsymbol{b}$ para alguna $\boldsymbol{b} \in \mathbb{R}^{k +1}$ y además $\boldsymbol{y} - X \boldsymbol{b} \in Im(X)^{\perp} = Ker(X^{T}).$

Por lo tanto, 

$$0 = \mathrm{X}^{T} (\boldsymbol{y} - \mathrm{X} \boldsymbol{b}) = \mathrm{X}^{T} X \boldsymbol{b}$$

y como $\mathrm{X}^{T} X$ es invertible por la izquierda, se tiene lo pedido.

**Coeficiente de determinación.**

<font color = 'Black'>
<font size = 3>

Al igual que hicimos con la **regresión lineal simple**, vamos a definir el **coeficiente de determinación** que es una manera de medir lo efectiva que es la **regresión**.

Como antes, este se define a través de la **identidad de suma de cuadrados**:

$$\underbrace{\sum_{i=1}^n(Y_i-\overline{Y})^2}_{SS_{Total}}=\underbrace{\sum_{i=1}^n(\hat{Y_i}-\overline{Y})^2}_{SS_{Regresión}}+\underbrace{\sum_{i=1}^n(Y_i-\hat{Y_i})^2}_{SS_{Error}}$$

Se define el **coeficiente de determinación $R^2$** en la **regresión** por el método de los **mínimos cuadrados** como $$R^2 = \frac{SS_{Regresión}}{SS_{Total}}$$

$R^2$ es una cantidad entre 0 y 1. Cuanto más próximo a 1 esté dicho coeficiente, más precisa será la recta de **regresión**.

---

<font color = 'DodgerBlue'>
<font size = 5>
    
**Comparación de modelos: coeficiente de determinación ajustado.**

<font color = 'Black'>
<font size = 3>

El **coeficiente de determinación** definido anteriormente aumenta si aumentamos el número de **variables independientes** $k$, incluso si estas aportan información redundante o poca información. Por ejemplo, si añadimos **variables** que son linealmente dependientes de las demás (**correlacionadas** con otras).

En **regresión lineal simple** el $R^2$ no tiene ningún problema. Pero, entre más **columnas explicativas** tengas, la $R^2$ va a mejorar, pero eso no implica que va ser mejor modelo. Ese es el inconveniente porque el $R^2$ se está inflando de manera artificial, no porque estés mejorando la **regresión**. Para eso tenemos el **coeficiente de regresión ajustado**.

Para evitar este problema, o para penalizar el aumento de **variables independientes** se usa en su lugar el **coeficiente de regresión ajustado:** $$R^2_{adj}=\frac{MS_T-MS_E}{MS_T}$$ donde $MS_T=\frac{SS_T}{n-1}$ y $MS_E=\frac{SS_E}{n-k-1}$. La relación entre ambos **coeficientes de determinación** es $$R^2_{adj}=1-(1-R^2)\frac{n-1}{n-k-1}$$

En general, $0\le R^2_{adj}<R^2\le1$, por lo el **coeficiente de determinación ajustado** es más difícil obtener un valor cercano a 1.

**Comparación de modelos: AIC y BIC.**

<font color = 'Black'>
<font size = 3>

Cuando comparamos dos **modelos lineales múltiples** para ver cuál es más adecuado que el otros, el **coeficiente de regresión ajustado** es uno de los métodos para tomar la decisión: siempre buscaremos aquel con **mayor $R^2$ ajustado**.

Existen más métodos como son el **AIC** o el **BIC** (**criterio de Akaike** y **criterio bayesiano**). En ambos, buscamos el de **menor información**.

<font color = 'DodgerBlue'>
<font size = 5>
    
**Índice de Akaike.**

<font color = 'Black'>
<font size = 3>

El **AIC** cuantifica cuánta información de $Y$ se pierde con el modelo y cuántas variables usamos. Concretamente, se define como $$AIC=n\ln\left(\frac{SS_E}{n}\right)+2k$$ y el modelo con **menor AIC** es el más adecuado.

En **Python**, para calcularlo usamos ``modelo.aic`` donde ``modelo`` es el modelo lineal obtenido..

<font color = 'DodgerBlue'>
<font size = 5>
    
**Índice bayesiano.**

<font color = 'Black'>
<font size = 3>

Análogo al **método de Akaike**, el **BIC** cuantifica cuánta información de $Y$ se pierde con el modelo y cuántas variables usamos. Concretamente, se define como $$BIC=n\ln\left(\frac{SS_E}{n}\right)+k\ln(n)$$ y el modelo con **menor BIC** es el más adecuado.

En **Python**, para calcularlo usamos ``modelo.bic`` donde ``modelo`` es el **modelo lineal** obtenido.

---

<font color = 'DodgerBlue'>
<font size = 5>
    
**Comparación de modelos en Python.**

<font color = 'Black'>
<font size = 3>

Supongamos que tienes $k$ variables explicativas. Entonces puedes formar $2^k-1$ **modelos lineales** diferentes con todas ellas. Para hallar el mejor, con **Python** podemos crear nuestra propia función para realizar la búsqueda, la cual encuentra el mejor mediante el **criterio de Akaike**.

Esta función compara todos los modelos posibles y se queda con el que tenga menor índice de Akaike.

```python
################################################################
########################    EN PYTHON    #######################
################################################################

import statsmodels.api as sm
import itertools

# Función para seleccionar el mejor modelo según el AIC
def stepwise_selection(data, target, predictors):
    """
    data: DataFrame con los datos.
    target: Nombre de la columna dependiente (Y).
    predictors: Lista de nombres de las columnas independientes (X).
    """
    best_aic = float('inf')
    best_model = None
    best_predictors = []

    # Generar todos los subconjuntos posibles de las variables independientes
    for k in range(1, len(predictors) + 1):
        for subset in itertools.combinations(predictors, k):
            X = sm.add_constant(data[list(subset)])  # Agregar el término constante
            Y = data[target]
            model = sm.OLS(Y, X).fit()
            aic = model.aic
            
            if aic < best_aic:  # Comparar el AIC
                best_aic = aic
                best_model = model
                best_predictors = list(subset)
    
    return best_model, best_predictors

# Uso del paso anterior
predictors = ['Variable_1', 'Variable_2', 'Variable_k']
target = 'Y'
best_model, best_predictors = stepwise_selection(datos, target, predictors)

# Imprimir resultados
print(f"Mejor conjunto de predictores: {best_predictors}")
print(f"AIC del mejor modelo: {best_model.aic}")
print(best_model.summary())

################################################################
```

````python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from plotnine import *
import itertools
````

````python
def stepwise_selection(data, target, predictors):
    """
    data: DataFrame con los datos.
    target: Nombre de la columna dependiente (Y).
    predictors: Lista de nombres de las columnas independientes (X).
    """
    best_aic = float('inf')
    best_model = None
    best_predictors = []

    # Generar todos los subconjuntos posibles de las variables independientes
    for k in range(1, len(predictors) + 1):
        for subset in itertools.combinations(predictors, k):
            X = data[list(subset)]  # Seleccionar las columnas del subset
            X = sm.add_constant(X)  # Agregar el término constante
            Y = data[target]

            try:
                model = sm.OLS(Y, X).fit()
                aic = model.aic

                if aic < best_aic:  # Comparar el AIC
                    best_aic = aic
                    best_model = model
                    best_predictors = list(subset)

            except Exception as e:
                print(f"Error al ajustar el modelo con el subset {subset}: {e}")

    return best_model, best_predictors
````

````python
# Uso del paso anterior
predictors = ['habitantes', 'ingresos', 'analfabetismo', 'asesinatos',
              'universitarios', 'heladas', 'area', 'densidad_pobl']
target = 'esp_vida'
````

````python
# Asegúrate de que 'muestra' tiene las columnas correctas y no hay valores NaN
muestra = muestra.dropna()
````

````python
best_model, best_predictors = stepwise_selection(muestra, target, predictors)
````

````python
# Ver los resultados
print("Mejores predictores:", best_predictors)
````

````python
print("Resumen del mejor modelo:")
print(best_model.summary())
````

**Significancia de los coeficientes**

<font color = 'Black'>
<font size = 3>

Nos interesa ahora saber cuáles de los **coeficientes** $\beta_1,\beta_2,...,\beta_k$ son significativamente diferentes de cero. 

Podemos analizar esta situación en dos pasos: 

* Por una parte, ¿existe alguno diferente de cero? Es decir, ¿nuestra regresión vale la pena? Para responder esto, bajo las hipótesis sobre los residuos expresadas anteriormente, se plantea la prueba

$$\left\{\begin{array}{l}H_0:\beta_1=\beta_2=...=\beta_k=0\\H_1:\mbox{ Existe }i\mbox{ con }\beta_i\neq0\end{array}\right.$$

El **estadístico de contraste** de esta prueba es una cierta $F$. Si el $p$-valor es pequeño, concluimos que la **regresión** vale la pena

* En caso de que la respuesta a la pregunta anterior sea positiva, ¿cuáles **coeficientes** son significativos?

Para esto, se plantean las $k$ pruebas

$$\left\{\begin{array}{l}H_0:\beta_i=0\\H_1:\beta_i\neq0\end{array}\right.$$

El **estadístico de contraste** de esta prueba es una cierta $t$. Si el $p$-valor es pequeño, concluimos que el **coeficiente** $\beta_i$ es siginificativamente diferente de 0 (por lo tanto los efectos de la **característica** $X_i$ son importantes).

**Estudio de predicciones.**

<font color = 'Black'>
<font size = 3>

Fijados valores concretos de $(x_1,x_2,...,x_n)$, digamos $(x_{01},x_{02},...,x_{0k})$, podemos considerar dos parámetros más a estudiar: el valor **medio** de la **variable aleatoria** $Y|_{x_{01},x_{02},...,x_{0k}}$ y el **valor estimado** $y_0=b_0+b_1x_{10}+...+b_kx_{k0}$ por la **regresión**.

Así, los **intervalos de confianza** al $100(1-\alpha)\%$ de **confianza** para ambos parámetros vienen dados por $$\left(\mu|_{\boldsymbol{x_0}}-t_{n-k-1,1-\frac{\alpha}{2}}S\sqrt{\boldsymbol{x_0^\prime}(\boldsymbol{\mathrm{X}}^T\boldsymbol{\mathrm{X}})^{-1}\boldsymbol{x_0}^{\prime T}},\mu|_{\boldsymbol{x_0}}+t_{n-k-1,1-\frac{\alpha}{2}}S\sqrt{\boldsymbol{x_0^\prime}(\boldsymbol{\mathrm{X}}^T\boldsymbol{\mathrm{X}})^{-1}\boldsymbol{x_0}^{\prime T}}\right)$$ y 
 
$$\left(\widehat{y_0}-t_{n-k-1,1-\frac{\alpha}{2}}S\sqrt{1+\boldsymbol{x_0^\prime}(\boldsymbol{\mathrm{X}}^T\boldsymbol{\mathrm{X}})^{-1}\boldsymbol{x_0}^{\prime T}},\widehat{y_0}+t_{n-k-1,1-\frac{\alpha}{2}}S\sqrt{1+\boldsymbol{x_0^\prime}(\boldsymbol{\mathrm{X}}^T\boldsymbol{\mathrm{X}})^{-1}\boldsymbol{x_0}^{\prime T}}\right)$$

- La **regresión lineal múltiple** es como en Machine Learning, cuando quieres predecir una característica $X$ a partir de $k$ columnas. Las técnicas de Machine Learning son generalizaciones de estas.
- La imagen de una transformación lineal siempre es un espacio vectorial.
- En espacios de Hilbert siempre puedes minimizar. Dado un punto en el espacio y un subespacio vectorial, siempre que sea cerrado, siempre puedes encontrar un vector cuya distancia sea mínima.
- La **Información de Akaike** es como la entropía del sistema.
- El modelo anidado permite hacer una regresión lineal y una máquina de soporte vectorial, y vas alimentando el modelo con la misma información. El problema de ellos es que son más tardados.
